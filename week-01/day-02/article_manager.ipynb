```python
import json

articles = []


def get_article():
    print("---- Article -----")
    title = str(input('Title: '))
    authors = str(input('Author(s) - separate with comma: '))
    while True:
        try:
            year = int(input('Year: '))
            break
        except ValueError:
            print("Please enter a valid number for year.")
    abstract = str(input('Abstract: '))
    keywords = str(input('Keyword(s) - separate with comma: '))

    return {
        'title': title,
        'authors': [name.strip() for name in authors.split(',')],
        'year': year,
        'abstract': abstract,
        'keywords': [kw.strip() for kw in keywords.split(',')],
    }


def validate_article(article):
    if article['title'] == '':
        return False
    if len(article['authors']) == 0:
        return False
    if article['year'] <= 1900 or article['year'] > 2070:
        print("Invalid year")
        return False
    if article['abstract'] == '':
        return False
    if len(article['keywords']) == 0 :
        return False
    return True


def display_articles(article_list):
    if not article_list:
        print("\nNo articles to display.")
        return

    print(f"\n{'='*50}")
    print(f"📚 Total Articles: {len(article_list)}")
    print(f"{'='*50}")

    for i, art in enumerate(article_list, 1):
        print(f"\n📄 Article {i}")
        print(f"{'-'*50}")
        print(f"Title    : {art['title']}")
        print(f"Author(s): {', '.join(art['authors'])}")
        print(f"Year     : {art['year']}")
        print(f"Abstract : {art['abstract']}")
        print(f"Keywords : {', '.join(art['keywords'])}")
    print(f"\n{'='*50}\n")


# --- main loop ---
while True:
    new_article = get_article()

    if validate_article(new_article):
        print("✅ Done")
        articles.append(new_article)
    else:
        print("❌ The data is missing/invalid. This article was not saved.")

    again = input('\nAdd another article? (y/n): ').strip().lower()
    if again != 'y':
        break

display_articles(articles)
```
