```python
def get_article():
    title = input('Insert the name of the article: ')
    authors = input('Insert authur(s): ')
    year = int(input('Insert year: '))
    abstract = input('Insert abstract: ')
    keywords = input('Insert keyword(s): ')

    # برگرداندن همهٔ داده‌ها در قالب یک دیکشنری
    return {
        'title': title,
        'authors': authors.split(", "),
        'year': year,
        'abstract': abstract,
        'keywords': keywords.split(", ")
    }


articles = []
while True:
    articles.append(get_article())
    more = input('آیا مقالهٔ دیگری اضافه می‌کنید؟ (y/n): ').lower()
    if more != 'y':
        break

# نمایش همهٔ مقالات
for i, art in enumerate(articles, 1):
    print(f"\nمقاله {i}:")
    print(art)
```
